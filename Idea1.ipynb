```python
from sentence_transformers import SentenceTransformer
'''
This is a pretrained sentence transformer model, instead of training BERT from scratch we can encode descriptions of sensors and
using cosine similarity we can find which old sensors the new one is most similar to.
'''
model = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

```python
examples = ["This is an example sentence","Another sentence which is just for demonstration"]
embeddings = model.encode(examples)
print(embeddings)
```

```python
from sentence_transformers import util
similarity = util.cos_sim(embeddings[0],embeddings[1])
score = similarity.item()
print("Similarity score ",score) 
'''
Interpretting the Score:
0.9 - 1: Near Identical: exact same meaning
0.7 - 0.89: Very Similar: About the same topic, same situation, same context just slight difference
0.4 - 0.69: Topical Similarity: Share the same broad theme/ topic but the actual meaning/action differ
0 - 0.39: Low to No Similarity
'''
```
